```python
from google.colab import drive
drive.mount('/content/drive')
import os

DATASET_DIR = "/content/drive/MyDrive/COMPUTER VISION_TI 2026/KELOMPOK 3/KLASIFIKASI CITRA DENGAN CNN/DATA MENTAH"

kelas = ["Laptop", "Buku", "Baju"]

ekstensi = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

total_semua = 0

for nama_kelas in kelas:

    folder_kelas = os.path.join(DATASET_DIR, nama_kelas)

    print(f"\n===== {nama_kelas.upper()} =====")

    total_kelas = 0

    for kondisi in ["Kondisi Ideal", "Kondisi Sulit"]:

        folder_kondisi = os.path.join(folder_kelas, kondisi)

        if not os.path.exists(folder_kondisi):
            print(f"{kondisi}: FOLDER TIDAK DITEMUKAN")
            continue

        jumlah = sum(
            1 for f in os.listdir(folder_kondisi)
            if f.lower().endswith(ekstensi)
        )

        print(f"{kondisi}: {jumlah} foto")

        total_kelas += jumlah

    print(f"TOTAL {nama_kelas}: {total_kelas}")

    total_semua += total_kelas

print("\n==============================")
print("TOTAL SELURUH FOTO:", total_semua)
print("==============================")
```
