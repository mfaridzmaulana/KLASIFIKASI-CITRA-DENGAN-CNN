```python

from google.colab import drive
drive.mount('/content/drive')

import os
import numpy as np

from sklearn.model_selection import train_test_split

# Lokasi dataset mentah
DATASET_DIR = "/content/drive/.shortcut-targets-by-id/1LoCHPgLL2xzbYllXiCpFE89B8TmW7Efs/COMPUTER VISION_TI 2026/kelompok 3/KLASIFIKASI CITRA DENGAN CNN/DATA MENTAH"

# Lokasi hasil preprocessing
OUTPUT_DIR = os.path.dirname(DATASET_DIR)

FILE_DATA = os.path.join(
    OUTPUT_DIR,
    "dataset_preprocessed.npz"
)

print("Persiapan pembagian dataset")
print("Lokasi file:", FILE_DATA)

if os.path.exists(FILE_DATA):
    print("File preprocessing ditemukan!")
else:
    print("File tidak ditemukan. Periksa lokasi penyimpanan!")
```

```python

# Membaca dataset hasil preprocessing
data = np.load(FILE_DATA)

X = data["X"]
y = data["y"]
kondisi_data = data["kondisi"]

print("DATASET BERHASIL DIBACA")
print("=" * 40)

print("Jumlah gambar:", len(X))
print("Bentuk gambar:", X.shape)
print("Jumlah label:", len(y))
print("Jumlah kondisi:", len(kondisi_data))
```

```python

# Membuat kombinasi label kelas dan kondisi
strata = np.array([
    f"{label}_{kondisi}"
    for label, kondisi in zip(y, kondisi_data)
])

# Membagi 70% training dan 30% sisa
X_train, X_sisa, y_train, y_sisa, k_train, k_sisa = train_test_split(
    X,
    y,
    kondisi_data,
    test_size=0.30,
    random_state=42,
    stratify=strata
)

# Membagi sisa menjadi validation dan testing
strata_sisa = np.array([
    f"{label}_{kondisi}"
    for label, kondisi in zip(y_sisa, k_sisa)
])

X_val, X_test, y_val, y_test, k_val, k_test = train_test_split(
    X_sisa,
    y_sisa,
    k_sisa,
    test_size=0.50,
    random_state=42,
    stratify=strata_sisa
)

print("PEMBAGIAN DATASET BERHASIL")
print("=" * 40)

print("Training   :", len(X_train))
print("Validation :", len(X_val))
print("Testing    :", len(X_test))

print("-" * 40)
print("Total      :", len(X_train) + len(X_val) + len(X_test))
```

```python

label_kelas = {
    0: "TAS",
    1: "BUKU GAMBAR",
    2: "LAPTOP"
}

def laporan_split(nama, label, kondisi):
    print("\n", nama)
    print("-" * 40)

    for angka, kelas in label_kelas.items():
        for kondisi_nama in ["KONDISI IDEAL", "KONDISI SULIT"]:

            jumlah = np.sum(
                (label == angka) &
                (kondisi == kondisi_nama)
            )

            print(f"{kelas} - {kondisi_nama}: {jumlah}")

laporan_split("TRAINING", y_train, k_train)
laporan_split("VALIDATION", y_val, k_val)
laporan_split("TESTING", y_test, k_test)
```

```python

SPLIT_FILE = os.path.join(
    OUTPUT_DIR,
    "dataset_split.npz"
)

np.savez_compressed(
    SPLIT_FILE,
    X_train=X_train,
    X_val=X_val,
    X_test=X_test,
    y_train=y_train,
    y_val=y_val,
    y_test=y_test,
    k_train=k_train,
    k_val=k_val,
    k_test=k_test
)

print("HASIL PEMBAGIAN DATA BERHASIL DISIMPAN")
print(SPLIT_FILE)
```
