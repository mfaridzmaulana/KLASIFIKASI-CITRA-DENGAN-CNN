```python

from google.colab import drive
drive.mount('/content/drive')

import os
import numpy as np
from PIL import Image

DATASET_DIR = "/content/drive/.shortcut-targets-by-id/1LoCHPgLL2xzbYllXiCpFE89B8TmW7Efs/COMPUTER VISION_TI 2026/kelompok 3/KLASIFIKASI CITRA DENGAN CNN/DATA MENTAH"

kelas_list = ["TAS", "BUKU GAMBAR", "LAPTOP"]
kondisi_list = ["KONDISI IDEAL", "KONDISI SULIT"]

label_kelas = {
    "TAS": 0,
    "BUKU GAMBAR": 1,
    "LAPTOP": 2
}

ukuran = (128, 128)

print("Persiapan preprocessing berhasil!")
print("Dataset:", DATASET_DIR)
```

```python

X = []
y = []
kondisi_data = []

ekstensi = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

for kelas in kelas_list:

    for kondisi in kondisi_list:

        folder = os.path.join(DATASET_DIR, kelas, kondisi)

        for nama_file in os.listdir(folder):

            if nama_file.lower().endswith(ekstensi):

                lokasi = os.path.join(folder, nama_file)

                gambar = Image.open(lokasi).convert("RGB")
                gambar = gambar.resize(ukuran)

                array_gambar = np.array(gambar)
                array_gambar = array_gambar / 255.0

                X.append(array_gambar)
                y.append(label_kelas[kelas])
                kondisi_data.append(kondisi)

X = np.array(X, dtype=np.float32)
y = np.array(y)
kondisi_data = np.array(kondisi_data)

print("PREPROCESSING SELESAI")
print("Jumlah gambar:", len(X))
print("Bentuk data gambar:", X.shape)
print("Bentuk label:", y.shape)
```

```python

print("LAPORAN HASIL PREPROCESSING")
print("=" * 40)

for kelas, label in label_kelas.items():
    print(kelas, ":", np.sum(y == label), "gambar")

print("-" * 40)

for kondisi in kondisi_list:
    print(kondisi, ":", np.sum(kondisi_data == kondisi))

print("-" * 40)
print("Total gambar:", len(X))
print("Ukuran gambar:", X.shape[1:])
print("Nilai piksel minimum:", X.min())
print("Nilai piksel maksimum:", X.max())
```

```python

OUTPUT_DIR = os.path.dirname(DATASET_DIR)

np.savez_compressed(
    os.path.join(OUTPUT_DIR, "dataset_preprocessed.npz"),
    X=X,
    y=y,
    kondisi=kondisi_data
)

print("Dataset preprocessing berhasil disimpan!")
```
